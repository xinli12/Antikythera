```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Function to convert radians to degrees
def rad_to_deg(rad):
    return rad * 180 / np.pi

# Load the CSV files
iso_df = pd.read_csv('results/bootstrap_results_isotropic.csv')
ideal_df = pd.read_csv('results/bootstrap_results_ideal.csv')

# Let's focus on the alpha parameters in the isotropic results and theta0 in ideal results
# First, identify all the alpha columns in the isotropic dataframe
alpha_cols = [col for col in iso_df['Parameter'] if 'alpha' in col]

# Create a new dataframe to store the results
results = []

# Process isotropic data
for param in iso_df['Parameter']:
    row = iso_df[iso_df['Parameter'] == param].iloc[0]
    
    # Convert alpha values to degrees if they're alpha parameters
    if 'alpha' in param:
        mean = rad_to_deg(row['Mean'])
        median = rad_to_deg(row['Median'])
        std = rad_to_deg(row['Std'])
        ci_68_lower = rad_to_deg(row['68.0%_lower'])
        ci_68_upper = rad_to_deg(row['68.0%_upper'])
        ci_90_lower = rad_to_deg(row['90.0%_lower'])
        ci_90_upper = rad_to_deg(row['90.0%_upper'])
        ci_95_lower = rad_to_deg(row['95.0%_lower'])
        ci_95_upper = rad_to_deg(row['95.0%_upper'])
        ci_99_lower = rad_to_deg(row['99.0%_lower'])
        ci_99_upper = rad_to_deg(row['99.0%_upper'])
    else:
        mean = row['Mean']
        median = row['Median']
        std = row['Std']
        ci_68_lower = row['68.0%_lower']
        ci_68_upper = row['68.0%_upper']
        ci_90_lower = row['90.0%_lower']
        ci_90_upper = row['90.0%_upper']
        ci_95_lower = row['95.0%_lower']
        ci_95_upper = row['95.0%_upper']
        ci_99_lower = row['99.0%_lower']
        ci_99_upper = row['99.0%_upper']
    
    # Calculate differences
    diff_median_68_lower = median - ci_68_lower
    diff_median_68_upper = ci_68_upper - median
    diff_median_90_lower = median - ci_90_lower
    diff_median_90_upper = ci_90_upper - median
    diff_median_95_lower = median - ci_95_lower
    diff_median_95_upper = ci_95_upper - median
    diff_median_99_lower = median - ci_99_lower
    diff_median_99_upper = ci_99_upper - median
    diff_mean_minus_std = mean - std
    diff_mean_plus_std = mean + std
    
    results.append({
        'Parameter': param,
        'Mean': mean,
        'Median': median,
        'Std': std,
        'Median-68%_lower': diff_median_68_lower,
        '68%_upper-Median': diff_median_68_upper,
        'Median-90%_lower': diff_median_90_lower,
        '90%_upper-Median': diff_median_90_upper,
        'Median-95%_lower': diff_median_95_lower,
        '95%_upper-Median': diff_median_95_upper,
        'Median-99%_lower': diff_median_99_lower,
        '99%_upper-Median': diff_median_99_upper,
        'Mean-Std': diff_mean_minus_std,
        'Mean+Std': diff_mean_plus_std
    })

results_df = pd.DataFrame(results)
results_df.to_csv('results/bs_final_isotropic.csv', index=False)
results_df
```

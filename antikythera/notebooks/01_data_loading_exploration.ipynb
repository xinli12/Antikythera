## Data Loading

We'll load the data from the CSV file containing the measured hole locations. Each hole has x and y coordinates in millimeters, and is associated with a specific section of the fragmented ring.

```python
# Load the data
df = pd.read_csv(data_file)

# Display the first few rows
print(f"Number of holes: {len(df)}")

# Count holes in each section
section_counts = df['Section ID'].value_counts().sort_index()
print("Number of holes in each section:")
print(section_counts)

df.head()
```
